```python
import torch
from transformers import pipeline, AutoModelForSeq2SeqLM, AutoTokenizer
model_name = "google/flan-t5-large"

model = AutoModelForSeq2SeqLM.from_pretrained(
    model_name,
    local_files_only=True
)
tokenizer = AutoTokenizer.from_pretrained(
    model_name,
    local_files_only=True
)
```

```python
from transformers import pipeline
pipe = pipeline(
    "text2text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=100, 
)
```

```python
#If not installed earlier get langchain-huggingface
#!pip install -U langchain-huggingface
from langchain_huggingface import HuggingFacePipeline
```

```python
llm = HuggingFacePipeline(pipeline=pipe)
```

```python
prompt_test = "what is the main cause of weather change, explain"
```

```python
response = llm.invoke(prompt_test)
```

```python
print(response)
```

```python
#if not already imported earlier
from langchain_core.prompts import PromptTemplate
# Define prompt template
template = """Question: {question}
Answer: Let's think step by step."""
prompt = PromptTemplate(template=template, input_variables=["question"])
```

```python
# Example questions
questions = [
    "Explain the concept of black holes in simple terms.",
    "What are the main causes of climate change, and how can we address them?",
    "Provide a brief overview of the history of artificial intelligence."
]
```

```python
#Option 1(which works with Older version or langchain)
#from langchain.chains import LLMChain
#Create LLMChain
#llm_chain = LLMChain(prompt=prompt, llm=llm)

# # Run model for each question
# for question in questions:
#     response = llm_chain.run(question)
#     print(f"Q: {question}\nA: {response}\n")
```

```python
#Chaining using langchain
#Option 2(newer version)
from langchain_core.runnables import RunnableSequence
# # RunnableSequence
chain = prompt | llm

for q in questions:
     print(f"\nQ: {q}")
     print(chain.invoke({"question": q}))

#For other approaches such as using functions, more on templates & passing prompts directly into Pipe
#refer: 0.1.3.1-Langchain_Formatting_Instruc_templates.ipynb
```

```python
pipe1 = pipeline(
    "text2text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=100, 
    torch_dtype=torch.float32,  # Uses lower precision for efficiency
    device=0 if torch.cuda.is_available() else -1  # Use GPU if available
)
```

```python
llm2 = HuggingFacePipeline(pipeline=pipe1)
```

```python
questions = [
    "Why is retrieval augmented generation used in AI"
]
```

```python
chain2 = prompt | llm2

for q in questions:
     print(f"\nQ: {q}")
     print(chain2.invoke({"question": q}))
```
