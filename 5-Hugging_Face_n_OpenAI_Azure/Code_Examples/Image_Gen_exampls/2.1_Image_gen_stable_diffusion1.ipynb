```python
import torch
from diffusers import StableDiffusionPipeline

# Load a pre-trained Stable Diffusion model
# "runwayml/stable-diffusion-v1-5" is one of the most popular checkpoints
model_id = "runwayml/stable-diffusion-v1-5"

# If you have a GPU, use it; otherwise, fallback to CPU (slow)
device = "cuda" if torch.cuda.is_available() else "cpu"

# Create the pipeline
pipe = StableDiffusionPipeline.from_pretrained(
    model_id,local_files_only=True,
    dtype=torch.float32 if device == "cuda" else torch.float32
)

pipe = pipe.to(device)


```

```python
# Your prompt
prompt = "A serene forest at dawn with deer, warm sunlight, soft mist, oil painting style"

# Generate image when GPU
#image = pipe(prompt, guidance_scale=7.5).images[0]

''' 
num_inference_steps → how many times the model refines the image
guidance_scale → how strongly the model follows your text prompt
|  Steps | Speed                    | Typical result       |
| -----: | ------------------------ | -------------------- |
|      5 | Very fast                | Poor / incomplete    |
|     10 | Fast                     | Rough                |
| **15** | **Good for CPU testing** | Reasonable           |
|  20–30 | Slower                   | Better refinement    |
|     50 | Much slower              | Usually more refined |
'''
# Generate image when CPU
image = pipe(
    prompt,
    guidance_scale=7.5,
    num_inference_steps=15,
    height=512,
    width=512
).images[0]

# Save image
image.save("forest_with_deer.png")

print("Image saved as forest_with_deer.png")
```
